### 2. Notação de Dirac (Bra-Ket)

A notação de Dirac é a ferramenta padrão utilizada para representar e manipular estados em espaços vetoriais complexos ($\mathbb{C}^n$). Ela simplifica a representação geométrica e algébrica dos sistemas quânticos através de dois elementos fundamentais:

* **Ket $|\psi\rangle$:** Representa um **vetor coluna**. É o objeto matemático básico que define o estado quântico de um sistema (como um *qubit*).

$$|\psi\rangle = \begin{bmatrix} z_1 \\ z_2 \\ \vdots \\ z_n \end{bmatrix}$$

* **Bra $\langle\psi|$:** Representa um **vetor linha**. Ele pertence ao espaço dual e é obtido diretamente a partir do Ket correspondente através da operação de **transposta conjugada** (também chamada de adjunto ou *adjoint*, denotada pelo símbolo $\dagger$).

$$\langle\psi| = |\psi\rangle^\dagger = \begin{bmatrix} z_1^* & z_2^* & \dots & z_n^* \end{bmatrix}$$

> **Nota Matemática:** O símbolo $*$ indica o **conjugado complexo** de um número. Lembre-se que se um elemento do vetor for $z = a + bi$, seu conjugado complexo será $z^* = a - bi$ (invertendo o sinal da parte imaginária). A operação adjunta $|\psi\rangle^\dagger$ combina a transposição geométrica (linha vira coluna) com a conjugação complexa.
